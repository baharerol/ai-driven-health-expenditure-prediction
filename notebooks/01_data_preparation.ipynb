```python
import pandas as pd
import numpy as np
import random
```

```python
# Ortak sabitler
years = list(range(2015, 2024))  # 9 yıl
months = list(range(1, 13))
mevsimsel_etki = {
    1: 1.10, 2: 1.05, 3: 1.00, 4: 0.95, 5: 0.90, 6: 0.85,
    7: 0.90, 8: 0.95, 9: 1.00, 10: 1.05, 11: 1.10, 12: 1.15
}
enflasyon_oranlari = {
    2015: 0.08, 2016: 0.10, 2017: 0.12, 2018: 0.15, 2019: 0.17,
    2020: 0.18, 2021: 0.19, 2022: 0.21, 2023: 0.24
}
```

```python
#Kurum Türü Bazlı Veri
kurum_turleri = ["Devlet", "Özel", "Üniversite"]
data_kurum = []

# Her ay için 100 gözlem
for year in years:
    for month in months:
        for kurum in kurum_turleri:
            for i in range(100):  # veri artırımı burada gerçekleşiyor
                seasonal = mevsimsel_etki[month]
                muracaat = int(np.random.normal(loc=150_000, scale=30_000) * seasonal * np.random.uniform(0.9, 1.1))
                kronik = round(np.random.normal(loc=0.18, scale=0.04), 2)
                kronik = min(max(kronik, 0.05), 0.35)  # 0–1 arası tut
                maliyet = round(np.random.normal(loc=300, scale=50) * (1 + enflasyon_oranlari[year]) * seasonal * np.random.uniform(0.9, 1.1), 2)
                toplam = round(muracaat * maliyet, 2)
                data_kurum.append([
                    year, month, kurum, muracaat,
                    kronik, enflasyon_oranlari[year],
                    maliyet, toplam
                ])

df_kurum_large = pd.DataFrame(data_kurum, columns=[
    "Yıl", "Ay", "Kurum Türü", "Müracaat Sayısı",
    "Kronik Hasta Oranı", "Sağlık Enflasyonu",
    "Kişi Başı Maliyet", "Toplam Harcama"
])

print(df_kurum_large.shape)
print(df_kurum_large.head())

df_kurum_large.to_csv("kurum_turu_large.csv", index=False)
```

```python
#Hizmet Türü Bazlı Veri
hizmet_turleri = ["Ayakta", "Yatarak", "Ameliyat"]
data_hizmet = []

for year in years:
    for month in months:
        for hizmet in hizmet_turleri:
            for _ in range(100):  # Her ay için 100 gözlem
                seasonal = mevsimsel_etki[month]
                muracaat = int(np.random.normal(loc=80_000, scale=20_000) * seasonal * np.random.uniform(0.9, 1.1))
                yasli_orani = round(np.random.normal(loc=0.15, scale=0.03), 2)
                yasli_orani = min(max(yasli_orani, 0.05), 0.30)
                hizmet_zorluk = {"Ayakta": 0.8, "Yatarak": 1.2, "Ameliyat": 1.6}[hizmet]
                birim_maliyet = round(np.random.normal(loc=400, scale=70) * hizmet_zorluk * (1 + enflasyon_oranlari[year]) * seasonal * np.random.uniform(0.9, 1.1), 2)
                toplam = round(muracaat * birim_maliyet, 2)
                data_hizmet.append([
                    year, month, hizmet, muracaat,
                    yasli_orani, enflasyon_oranlari[year],
                    birim_maliyet, toplam
                ])

df_hizmet_large = pd.DataFrame(data_hizmet, columns=[
    "Yıl", "Ay", "Hizmet Türü", "Müracaat Sayısı",
    "Yaşlı Nüfus Oranı", "Sağlık Enflasyonu",
    "Birim Maliyet", "Toplam Harcama"
])

df_hizmet_large.to_csv("hizmet_turu_large.csv", index=False)
```

```python
#Reçete Verileri
data_recete = []

for year in years:
    for month in months:
        for _ in range(100):
            seasonal = mevsimsel_etki[month]
            recete_sayisi = int(np.random.normal(loc=2_000_000, scale=300_000) * seasonal * np.random.uniform(0.9, 1.1))
            ort_bedel = round(np.random.normal(loc=120, scale=20) * (1 + enflasyon_oranlari[year]) * seasonal * np.random.uniform(0.9, 1.1), 2)
            toplam_ilac = round(recete_sayisi * ort_bedel, 2)
            marka_orani = round(np.random.normal(loc=0.6, scale=0.1), 2)
            marka_orani = min(max(marka_orani, 0.3), 0.9)
            data_recete.append([
                year, month, recete_sayisi,
                ort_bedel, toplam_ilac, marka_orani,
                enflasyon_oranlari[year]
            ])

df_recete_large = pd.DataFrame(data_recete, columns=[
    "Yıl", "Ay", "Reçete Sayısı", "Ortalama Reçete Bedeli",
    "Toplam İlaç Harcaması", "Markalı İlaç Oranı", "Sağlık Enflasyonu"
])

df_recete_large.to_csv("recete_verisi_large.csv", index=False)

print("✔ Veriler başarıyla oluşturuldu ve CSV dosyalarına kaydedildi.")
```
